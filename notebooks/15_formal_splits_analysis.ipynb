## 3. Eventos radar

```python
ev=pd.read_parquet(OUT/'split_event_balance.parquet')
display(ev)
plt.figure(figsize=(8,4))
plt.bar(ev.split,ev.timestamp_event_rate_ge_45)
plt.ylabel('P(timestamp >=45 dBZ)')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 4. Regimes

```python
rb=pd.read_parquet(OUT/'split_regime_balance.parquet')
for space in rb.regime_space.unique():
    t=rb[rb.regime_space.eq(space)].pivot(index='regime',columns='split',values='prevalence')
    display(t)
```

## 5. Predictor shift vs train

```python
ps=pd.read_parquet(OUT/'split_predictor_shift_vs_train.parquet')
for split in ['validation','test_primary','test_stress_ood']:
    t=ps[ps.evaluation_split.eq(split)].copy()
    t['abs_smd']=t.standardized_mean_difference.abs()
    display(t.sort_values('abs_smd',ascending=False).head(12))
```
