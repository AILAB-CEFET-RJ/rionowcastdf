# CorrDiff - Fase 16 - Baselines Formais

Análise dos baselines no split congelado da Fase 15.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

OUT = Path('../analysis_outputs/16_baselines')
```

## Histórico de treino

```python
hist = pd.read_parquet(OUT/'training_history.parquet')
display(hist)
for model, g in hist.groupby('model'):
    plt.figure(figsize=(7,4))
    plt.plot(g.epoch, g.train_loss, marker='o', label='train')
    plt.plot(g.epoch, g.val_loss, marker='o', label='validation')
    plt.title(model)
    plt.xlabel('epoch')
    plt.ylabel('loss')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## Métricas pontuais

```python
point = pd.read_parquet(OUT/'point_metrics.parquet')
display(point.sort_values(['split','domain','baseline']))
```

## Threshold skill

```python
thr = pd.read_parquet(OUT/'threshold_metrics.parquet')
display(thr[(thr.cohort=='full_split') & (thr.threshold_dbz.isin([30,40,45]))].sort_values(['split','threshold_dbz','baseline']))
```

## FSS

```python
fss = pd.read_parquet(OUT/'fss_metrics.parquet')
display(fss[(fss.cohort=='full_split') & (fss.threshold_dbz.isin([30,40,45]))].sort_values(['split','threshold_dbz','nominal_support_km','baseline']))
```

## Máximo por patch / extremos

```python
patch = pd.read_parquet(OUT/'patch_max_metrics.parquet')
display(patch.sort_values(['split','condition','baseline']))
```

## Métricas probabilísticas

```python
prob = pd.read_parquet(OUT/'probabilistic_metrics.parquet')
display(prob.sort_values(['split','metric','threshold_dbz']))
```

## Persistência em coorte comum

```python
common = thr[thr.cohort=='persistence_common']
display(common[common.threshold_dbz.isin([30,40,45])].sort_values(['split','threshold_dbz','baseline']))
```
