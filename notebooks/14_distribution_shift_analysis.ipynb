# CorrDiff - Fase 14 - Distribution Shift

Coverage, shift de predictors, composição de regimes e shift condicional do radar.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

OUT = Path('../analysis_outputs/14_distribution_shift')
summary = json.loads((OUT/'analysis_summary.json').read_text())
summary
```

## 1. Coverage por ano

```python
cov = pd.read_parquet(OUT/'coverage_by_year.parquet')
display(cov)
plt.figure(figsize=(9,4))
plt.plot(cov.year_utc, cov.availability_ratio, marker='o')
plt.xlabel('Ano')
plt.ylabel('Availability ratio')
plt.tight_layout()
plt.show()
```

## 2. Predictor shift

```python
pred = pd.read_parquet(OUT/'predictor_shift_metrics.parquet')
recent = pred[pred.comparison_id.eq('RECENT_2023_2024_vs_pre2023')].copy()
recent['abs_smd'] = recent.standardized_mean_difference.abs()
display(recent.sort_values('abs_smd', ascending=False).head(20))
```

## 3. Shift em anomalias mês×hora

```python
anom = pd.read_parquet(OUT/'predictor_shift_anomaly_metrics.parquet')
recent = anom[anom.comparison_id.eq('RECENT_2023_2024_vs_pre2023')].copy()
recent['abs_smd'] = recent.standardized_mean_difference.abs()
display(recent.sort_values('abs_smd', ascending=False).head(20))
```

## 4. Shift de composição dos regimes

```python
reg = pd.read_parquet(OUT/'regime_composition_shift.parquet')
display(reg)
```

## 5. Radar marginal

```python
event = pd.read_parquet(OUT/'radar_event_shift.parquet')
display(event[event.event_id.eq('ge_45')])
```

## 6. >=45 dBZ dentro dos regimes

```python
cond = pd.read_parquet(OUT/'conditional_event_shift_by_regime.parquet')
t = cond[(cond.comparison_id.eq('RECENT_2023_2024_vs_pre2023')) & (cond.event_id.eq('ge_45'))]
display(t)
```

## 7. Decomposição composition vs within-regime

```python
dec = pd.read_parquet(OUT/'event_rate_decomposition.parquet')
t = dec[(dec.comparison_id.eq('RECENT_2023_2024_vs_pre2023')) & (dec.event_id.eq('ge_45'))]
display(t)
```

## Regra de interpretação

Interprete target shift somente depois do coverage. Composition e within-regime são decomposições estatísticas, não causais.
